## 1.2 Ridge и Lasso с подбором alpha по кросс-валидации

Замените `LinearRegression` в том же пайплайне на `Ridge`, а затем на `Lasso(max_iter=20000)`.
Для каждой модели подберите `alpha` из сетки `alphas` по среднему $R^2$ на 5-кратной кросс-валидации **на обучающей выборке**.

Используйте `cross_val_score(model, X, y, cv=cv, scoring="r2")` — функция сама делит данные на фолды, обучает модель на каждом разбиении и возвращает массив оценок.

Постройте график «среднее $R^2$ на CV в зависимости от alpha» для обеих моделей (ось X — логарифмическая).

```python
alphas = np.logspace(-1, 3, 13)
cv_reg = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

ridge_cv_scores = []  # средний R2 на CV для каждого alpha
lasso_cv_scores = []

# <ВАШ КОД>
for alpha in alphas:
  # Ridge
  ridge_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Ridge(alpha=alpha))
  r2_ridge = cross_val_score(ridge_pipe, X_tr_d, y_tr_d, cv=cv_reg, scoring="r2").mean()
  ridge_cv_scores.append(r2_ridge)

  # Lasso
  lasso_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Lasso(alpha=alpha, max_iter=20_000))
  r2_lasso = cross_val_score(lasso_pipe, X_tr_d, y_tr_d, cv=cv_reg, scoring="r2").mean()
  lasso_cv_scores.append(r2_lasso)

best_alpha_ridge = alphas[np.argmax(ridge_cv_scores)]  # <ВАШ КОД>
best_alpha_lasso = alphas[np.argmax(lasso_cv_scores)]  # <ВАШ КОД>

# график
# <ВАШ КОД>
plt.figure(figsize=(10, 6))
plt.plot(alphas, ridge_cv_scores, label="Ridge", marker='o')
plt.plot(alphas, lasso_cv_scores, label='Lasso', marker='o')
plt.xscale('log')
plt.xlabel('alpha (логарифмическая шкала)')
plt.ylabel('Среднее R2 на кросс-валидации')
plt.title('Подбор гиперпараметра Alpha')
plt.grid(True, which='both', ls='--')
plt.legend()
plt.show()
```

```python
print(best_alpha_ridge)
print(best_alpha_lasso)
```

Обучите Ridge и Lasso с лучшими `alpha` на всей обучающей выборке и посчитайте $R^2$ на тесте.
Для Lasso посчитайте, сколько коэффициентов из 285 осталось ненулевыми.

```python
ridge_best = best_alpha_ridge  # <ВАШ КОД>
lasso_best = best_alpha_lasso  # <ВАШ КОД>

# Ridge
ridge_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Ridge(alpha=ridge_best))  # <ВАШ КОД>
ridge_pipe.fit(X_tr_d, y_tr_d)
r2_test_ridge=ridge_pipe.score(X_te_d, y_te_d)
# Lasso
lasso_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Lasso(alpha=lasso_best, max_iter=20_000))  # <ВАШ КОД>
lasso_pipe.fit(X_tr_d, y_tr_d)
r2_test_lasso = lasso_pipe.score(X_te_d, y_te_d)

n_nonzero_lasso = np.sum(lasso_pipe.named_steps['lasso'].coef_!=0)  # <ВАШ КОД>

print(f"Ridge (alpha={best_alpha_ridge:.2f}): R2 test = {r2_test_ridge:.3f}")
print(f"Lasso (alpha={best_alpha_lasso:.2f}): R2 test = {r2_test_lasso:.3f}, ненулевых коэффициентов: {n_nonzero_lasso}")
```

```python
# Проверка (не изменяйте)
assert len(ridge_cv_scores) == len(alphas) and len(lasso_cv_scores) == len(alphas)
assert r2_test_ridge > 0.35, "Ridge с подобранным alpha должен заметно превзойти LinearRegression"
assert r2_test_lasso > 0.45, "Lasso с подобранным alpha должен давать R2 > 0.45"
assert 0 < n_nonzero_lasso < 285
print("OK")
```

**Вопрос 1.** Почему LinearRegression на полиномах 3-й степени дала отрицательный $R^2$ на тесте, а Ridge и Lasso — нет? Чем отличается то, как Ridge и Lasso «борются» с лишними признаками?

**Ответ:**   
LinearRegression на полиномах 3-й степени дала отрицательный $R^2$ на тесте из-за переобучения и мультиколлинеарности. После добавления полиномиальных признаков, общее количество признаков увеличивается. LinearRegression пытается точно подстроиться под обучающую выборку, у неё нет штрафа за сложность (величину весов). По итогу модель в условиях избытка признаков и их сильной взаимосвязи (мультиколлинеарности) подбирает большие по модулю коэффициенты, которые компенсируют друг друга на тренировочных данных, но приводят к ошибкам на тестовых данных.  
В отличие от LinearRegression, Ridge и Lasso используют регуляризацию. Функция потерь штрафует модель за сложность (величину весов), что не даёт модели переобучиться.

Ridge добавляет к функции потерь квадраты весов, плавно уменьшая их значение и предотвращая переобучение.  
Lasso добавляет к функции потерь сумму модулей весов. Она может обратить часть весов в 0 у наименее важных или дублирующих признаков - таким образом модель сама отбирает признаки.
