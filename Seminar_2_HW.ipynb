# Часть 1. Регуляризация (2 балла)

Датасет `diabetes` встроен в sklearn: 442 пациента, 10 признаков (возраст, пол, индекс массы тела `bmi`, давление `bp` и шесть показателей анализа крови `s1`–`s6`). Целевая переменная — количественная мера прогрессирования диабета через год после обследования.

Признаки в датасете уже отцентрированы и отнормированы.

```python
X_diab, y_diab = load_diabetes(return_X_y=True, as_frame=True)
X_tr_d, X_te_d, y_tr_d, y_te_d = train_test_split(X_diab, y_diab, test_size=0.25,
                                                  random_state=RANDOM_STATE)
X_tr_d.shape, X_te_d.shape
```

## 1.1 Переобучение полиномиальной модели

Постройте модель `make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), LinearRegression())`.
Посчитайте $R^2$ на обучающей и тестовой выборках.

```python
poly_lr = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), LinearRegression())  # <ВАШ КОД>

poly_lr.fit(X_tr_d, y_tr_d)

r2_train_lr = poly_lr.score(X_tr_d, y_tr_d)  # <ВАШ КОД>
r2_test_lr = poly_lr.score(X_te_d, y_te_d)   # <ВАШ КОД>
print(f"R2 train: {r2_train_lr:.3f}, R2 test: {r2_test_lr:.3f}")
```

## 1.2 Ridge и Lasso с подбором alpha по кросс-валидации

Замените `LinearRegression` в том же пайплайне на `Ridge`, а затем на `Lasso(max_iter=20000)`.
Для каждой модели подберите `alpha` из сетки `alphas` по среднему $R^2$ на 5-кратной кросс-валидации **на обучающей выборке**.

Используйте `cross_val_score(model, X, y, cv=cv, scoring="r2")` — функция сама делит данные на фолды, обучает модель на каждом разбиении и возвращает массив оценок.

Постройте график «среднее $R^2$ на CV в зависимости от alpha» для обеих моделей (ось X — логарифмическая).

```python
alphas = np.logspace(-1, 3, 13)
cv_reg = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

ridge_cv_scores = []  # средний R2 на CV для каждого alpha
lasso_cv_scores = []

# <ВАШ КОД>
for alpha in alphas:
  # Ridge
  ridge_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Ridge(alpha=alpha))
  r2_ridge = cross_val_score(ridge_pipe, X_tr_d, y_tr_d, cv=cv_reg, scoring="r2").mean()
  ridge_cv_scores.append(r2_ridge)

  # Lasso
  lasso_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Lasso(alpha=alpha, max_iter=20_000))
  r2_lasso = cross_val_score(lasso_pipe, X_tr_d, y_tr_d, cv=cv_reg, scoring="r2").mean()
  lasso_cv_scores.append(r2_lasso)

best_alpha_ridge = alphas[np.argmax(ridge_cv_scores)]  # <ВАШ КОД>
best_alpha_lasso = alphas[np.argmax(lasso_cv_scores)]  # <ВАШ КОД>

# график
# <ВАШ КОД>
plt.figure(figsize=(10, 6))
plt.plot(alphas, ridge_cv_scores, label="Ridge", marker='o')
plt.plot(alphas, lasso_cv_scores, label='Lasso', marker='o')
plt.xscale('log')
plt.xlabel('alpha (логарифмическая шкала)')
plt.ylabel('Среднее R2 на кросс-валидации')
plt.title('Подбор гиперпараметра Alpha')
plt.grid(True, which='both', ls='--')
plt.legend()
plt.show()
```

```python
print(best_alpha_ridge)
print(best_alpha_lasso)
```

Обучите Ridge и Lasso с лучшими `alpha` на всей обучающей выборке и посчитайте $R^2$ на тесте.
Для Lasso посчитайте, сколько коэффициентов из 285 осталось ненулевыми.

```python
ridge_best = best_alpha_ridge  # <ВАШ КОД>
lasso_best = best_alpha_lasso  # <ВАШ КОД>

# Ridge
ridge_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Ridge(alpha=ridge_best))  # <ВАШ КОД>
ridge_pipe.fit(X_tr_d, y_tr_d)
r2_test_ridge=ridge_pipe.score(X_te_d, y_te_d)
# Lasso
lasso_pipe = make_pipeline(PolynomialFeatures(3, include_bias=False), StandardScaler(), Lasso(alpha=lasso_best, max_iter=20_000))  # <ВАШ КОД>
lasso_pipe.fit(X_tr_d, y_tr_d)
r2_test_lasso = lasso_pipe.score(X_te_d, y_te_d)

n_nonzero_lasso = np.sum(lasso_pipe.named_steps['lasso'].coef_!=0)  # <ВАШ КОД>

print(f"Ridge (alpha={best_alpha_ridge:.2f}): R2 test = {r2_test_ridge:.3f}")
print(f"Lasso (alpha={best_alpha_lasso:.2f}): R2 test = {r2_test_lasso:.3f}, ненулевых коэффициентов: {n_nonzero_lasso}")
```

```python
# Проверка (не изменяйте)
assert len(ridge_cv_scores) == len(alphas) and len(lasso_cv_scores) == len(alphas)
assert r2_test_ridge > 0.35, "Ridge с подобранным alpha должен заметно превзойти LinearRegression"
assert r2_test_lasso > 0.45, "Lasso с подобранным alpha должен давать R2 > 0.45"
assert 0 < n_nonzero_lasso < 285
print("OK")
```

**Вопрос 1.** Почему LinearRegression на полиномах 3-й степени дала отрицательный $R^2$ на тесте, а Ridge и Lasso — нет? Чем отличается то, как Ridge и Lasso «борются» с лишними признаками?

**Ответ:**   
LinearRegression на полиномах 3-й степени дала отрицательный $R^2$ на тесте из-за переобучения и мультиколлинеарности. После добавления полиномиальных признаков, общее количество признаков увеличивается. LinearRegression пытается точно подстроиться под обучающую выборку, у неё нет штрафа за сложность (величину весов). По итогу модель в условиях избытка признаков и их сильной взаимосвязи (мультиколлинеарности) подбирает большие по модулю коэффициенты, которые компенсируют друг друга на тренировочных данных, но приводят к ошибкам на тестовых данных.  
В отличие от LinearRegression, Ridge и Lasso используют регуляризацию. Функция потерь штрафует модель за сложность (величину весов), что не даёт модели переобучиться.

Ridge добавляет к функции потерь квадраты весов, плавно уменьшая их значение и предотвращая переобучение.  
Lasso добавляет к функции потерь сумму модулей весов. Она может обратить часть весов в 0 у наименее важных или дублирующих признаков - таким образом модель сама отбирает признаки.

## 2.1 Подготовка признаков

Соберите матрицу признаков `X` и целевую переменную `y`:

Разбиение на обучающую и тестовую выборки (не изменяйте). Параметр `stratify=y` сохраняет долю отказов одинаковой в обеих частях — при редком классе это важно.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y,
                                                    random_state=RANDOM_STATE)
y_train.mean(), y_test.mean()
```

## 2.2 Базовые модели и метрики

Обучите на `X_train`:
* `DummyClassifier()` — всегда предсказывает самый частый класс;
* `make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))`.

Для каждой модели посчитайте на `X_test` accuracy, precision, recall, ROC-AUC и PR-AUC (`average_precision_score`; для ROC-AUC и PR-AUC нужны вероятности из `predict_proba`, а не метки классов).
Сложите результаты в таблицу `metrics` (строки — модели `"dummy"` и `"logreg"`, столбцы — `"accuracy"`, `"precision"`, `"recall"`, `"roc_auc"`, `"pr_auc"`). Выведите матрицу ошибок логистической регрессии.

```python
# DummyClassifier
dummy = DummyClassifier()   # <ВАШ КОД>
dummy.fit(X_train, y_train)

# LogisticRegression
logreg = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))  # <ВАШ КОД>
logreg.fit(X_train, y_train)

# Предсказания на тестовой выборке
y_pred_dummy = dummy.predict(X_test)
y_proba_dummy = dummy.predict_proba(X_test)[:, 1]

y_pred_logreg = logreg.predict(X_test)
y_proba_logreg = logreg.predict_proba(X_test)[:, 1]

# собираем метрики в словарь
metrics_dict = {
    "dummy": {
        "accuracy": accuracy_score(y_test, y_pred_dummy),
        "precision": precision_score(y_test, y_pred_dummy, zero_division=0),
        "recall": recall_score(y_test, y_pred_dummy),
        "roc_auc": roc_auc_score(y_test, y_proba_dummy),
        "pr_auc": average_precision_score(y_test, y_proba_dummy),
    },
    "logreg": {
        "accuracy": accuracy_score(y_test, y_pred_logreg),
        "precision": precision_score(y_test, y_pred_logreg),
        "recall": recall_score(y_test, y_pred_logreg),
        "roc_auc": roc_auc_score(y_test, y_proba_logreg),
        "pr_auc": average_precision_score(y_test, y_proba_logreg),
    },
}

metrics = pd.DataFrame.from_dict(metrics_dict, orient="index")  # <ВАШ КОД>
metrics
```

```python
# Проверка (не изменяйте)
assert set(metrics.index) == {"dummy", "logreg"}
assert set(metrics.columns) >= {"accuracy", "precision", "recall", "roc_auc", "pr_auc"}
assert metrics.loc["dummy", "accuracy"] > 0.96 and metrics.loc["dummy", "recall"] == 0
assert metrics.loc["logreg", "roc_auc"] > 0.85
assert metrics.loc["logreg", "pr_auc"] < metrics.loc["logreg", "roc_auc"]
print("OK")
```

**Вопрос 2.** У DummyClassifier accuracy около 0.97. Почему эта модель бесполезна для задачи, и почему PR-AUC здесь информативнее, чем accuracy и ROC-AUC?

**Ответ:**  
DummyClassifier бесполезна для задчи, потому что она не способна обнаружить ни одной реальной неисправности (её Recall равен 0). Из-за дисбаланса классов (станок чаще работает штатно, чем ломается) она всегда будет предсказывать самый частый класс - "станок работает штатно".  

PR-AUC здесь информативнее, чем accuracy и ROC-AUC, потому что она наиболее объективна, чем другие.  Для DummyClassifier метрика PR-AUC будет равна базовой доле отказов в данных.  
Accuracy бесполезна, так как при сильном дисбалансе классов она показывает не качество работы модели, а просто пропорцию преобладающего класса в датасете.   
ROC-AUC может быть обманчиво оптимистичным. Метрика ROC-AUC учитывает долю ложноположительных ответов (FPR), которая рассчитывается относительно общего числа здоровых объектов (True Negatives). Поскольку исправного оборудования в разы больше, даже если модель совершит много ложных тревог, огромный знаменатель TN сгладит ошибку, и ROC-AUC покажет высокий результат, скрывая низкое качество прогноза.   
PR-AUC наиболее объективна, так как она полностью сфокусирована на целевом классе (отказах). В её расчете участвуют только True Positives, False Positives и False Negatives. Она оценивает, насколько точно модель находит именно поломки (Recall) и как часто при этом ложно срабатывает (Precision).

## 3.1 Сравнение моделей

Сравните четыре модели по PR-AUC на 5-кратной **стратифицированной** кросс-валидации на обучающей выборке (`scoring="average_precision"`):

## 3.3 Признаки из физики процесса

Отказы в этом датасете вызваны физическими причинами: перегрев (мала разница температур при низкой скорости), отказ по мощности (мощность вне допустимого диапазона), перегрузка (износ × момент слишком велик).
Напишите функцию `add_physics_features`, которая добавляет к таблице три признака:
* `power` — механическая мощность, Вт: $P = M \cdot \omega$, где $\omega = 2\pi \cdot \text{rpm} / 60$;
* `temp_diff` — разность `process_temp - air_temp`;
* `strain` — произведение `tool_wear * torque`.

Сравните PR-AUC на CV для логистической регрессии и XGBoost без новых признаков и с ними.

```python
def add_physics_features(data):
    data = data.copy()
    # <ВАШ КОД>
    # Добавление признаков
    data['power'] = data["torque"] * (2 * np.pi * data["rpm"] / 60)
    data["temp_diff"] = data["process_temp"] - data["air_temp"]
    data["strain"] = data["tool_wear"] * data["torque"]

    return data


X_train_fe = add_physics_features(X_train)
X_test_fe = add_physics_features(X_test)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

models_fe = {
    "logreg": make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
    "xgb": XGBClassifier(random_state=RANDOM_STATE, n_jobs=-1)
}


fe_results = pd.DataFrame(columns=["base", "physics"], index=["logreg", "xgb"])  # таблица: строки "logreg", "xgb"; столбцы "base", "physics" (средний PR-AUC на CV)
# <ВАШ КОД>
for name, model in models_fe.items():
    # Считаем PR-AUC на базовых признаках
    score_base = cross_val_score(model, X_train, y_train, cv=cv, scoring="average_precision").mean()
    # Считаем PR-AUC на новых физических признаках
    score_physics = cross_val_score(model, X_train_fe, y_train, cv=cv, scoring="average_precision").mean()

    fe_results.loc[name, "base"] = score_base
    fe_results.loc[name, "physics"] = score_physics

fe_results
```

```python
# Проверка (не изменяйте)
assert {"power", "temp_diff", "strain"} <= set(X_train_fe.columns)
assert abs(X_train_fe["power"].median() - 6100) < 300, "Проверьте формулу мощности"
assert fe_results.loc["xgb", "physics"] > fe_results.loc["xgb", "base"] + 0.03
print("OK")
```

**Вопрос 4.** XGBoost заметно выиграл от новых признаков, а логистическая регрессия — почти нет. Почему? Подсказка: отказ по мощности случается, когда мощность **либо слишком мала, либо слишком велика**.

**Ответ:**  
XGBoost выиграл от новых признаков, потому что он выделяет зоны "слишком малой" и "слишком высокой" мощности, что привело к заметному увеличению метрики PR-AUC.   
XGBoost построен на основе решающих деревьев, которые разбивают пространство признаков на кусочно-постоянные интервалы. Градиентный бустинг легко находит сложные, нелинейные и немонотонные паттерны в данных. Для XGBoost не составляет труда построить комбинацию разветвлений (порога вида: power < min_threshold или power > max_threshold), тем самым идеально выделив аномальные зоны "слишком малой" и "слишком высокой»"мощности, что и привело к заметному скачку метрики PR-AUC.

Логистическая регрессия почти не выграла от новых признаков из-за своей линейности.    
Логистическая регрессия - это строго линейный классификатор. Она способна провести в пространстве признаков только прямую линию (гиперплоскость). Подсказка указывает, что отказ по мощности имеет U-образную (или двустороннюю) природу: станок ломается, если мощность либо слишком мала, либо слишком велика. Линейная модель физически не способна разделить данные по такому правилу. Для неё увеличение мощности может означать либо только рост вероятности поломки, либо только её снижение. В результате логистическая регрессия просто не смогла извлечь выгоду из нового признака power.

## 4.1 Optuna

Напишите функцию `objective(trial)`, которая предлагает гиперпараметры XGBoost и возвращает средний PR-AUC на кросс-валидации `cv` по `X_train_fe`. Подбирайте как минимум `n_estimators` (100–600), `max_depth` (2–8), `learning_rate` (0.01–0.3, логарифмическая шкала), `subsample` (0.6–1.0), `min_child_weight` (1–10).
Запустите 30 испытаний.

```python
import optuna
optuna.logging.set_verbosity(optuna.logging.WARNING)


def objective(trial):
    # <ВАШ КОД>
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 600),
        "max_depth": trial.suggest_int("max_depth", 2, 8),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
        "subsample": trial.suggest_float("subsample", 0.6, 1.0),
        "min_child_weight": trial.suggest_int("min_child_weight", 1, 10),
        "random_state": RANDOM_STATE,
        "n_jobs": -1,
    }

    model = XGBClassifier(**params)

    scores = cross_val_score(model, X_train_fe, y_train, cv=cv, scoring="average_precision")

    return scores.mean()


study = optuna.create_study(direction="maximize", sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE))
study.optimize(objective, n_trials=30, show_progress_bar=True)
print("Лучший PR-AUC на CV:", study.best_value)
print("Лучшие параметры:", study.best_params)
```

## 4.3 Выбор порога по стоимости ошибок

Модель выдаёт вероятность отказа, а решение «останавливать станок на обслуживание или нет» принимается по порогу. Пусть **пропущенный отказ стоит 20 условных единиц, а ложная тревога — 1**.

1. Возьмите XGBoost с лучшими параметрами Optuna и получите **out-of-fold** вероятности на обучающей выборке: `cross_val_predict(model, X_train_fe, y_train, cv=cv, method="predict_proba")`. Каждая вероятность в них предсказана моделью, которая этот объект не видела.
2. Подберите по ним порог из `thresholds`, минимизирующий суммарную стоимость ошибок.
3. Обучите модель на всей обучающей выборке и **один раз** оцените её на тесте: PR-AUC и стоимость ошибок при подобранном пороге и при пороге 0.5.

```python
COST_FN, COST_FP = 20, 1
thresholds = np.linspace(0.005, 0.995, 199)

# Ф-ия вычисления стоимости ошибок
def total_cost(y_true, proba, threshold):
    # <ВАШ КОД>
    y_pred = (proba >= threshold).astype(int)
    fn = np.sum((y_true == 1) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))

    return fn * COST_FN + fp * COST_FP

best_xgb = XGBClassifier(**study.best_params, random_state=RANDOM_STATE, n_jobs=-1)
oof_proba = cross_val_predict(best_xgb, X_train_fe, y_train, cv=cv, method="predict_proba")[:, 1]

# Получаем out-of-fold вероятности на обучающей выборке
best_xgb = XGBClassifier(**grid.best_params_, random_state=RANDOM_STATE, n_jobs=-1)
oof_proba = cross_val_predict(best_xgb, X_train_fe, y_train, cv=cv, method="predict_proba")[:, 1]

# Подбираем порог на основе out-of-fold вероятностей
costs = [total_cost(y_train, oof_proba, t) for t in thresholds]
best_threshold = thresholds[np.argmin(costs)]  # <ВАШ КОД>

# Обучаем финальную модель на всей выборке
final_model = XGBClassifier(**grid.best_params_, random_state=RANDOM_STATE, n_jobs=-1)
final_model.fit(X_train_fe, y_train)

# Получаем вероятности на тестовой выборке
test_proba = final_model.predict_proba(X_test_fe)[:, 1]

test_pr_auc = average_precision_score(y_test, test_proba)
cost_best = total_cost(y_test, test_proba, best_threshold)
cost_05 = total_cost(y_test, test_proba, 0.5)
cost_never = total_cost(y_test, test_proba, 1.01)  # никогда не останавливать станок

print(f"PR-AUC на тесте: {test_pr_auc:.3f}")
print(f"Порог {best_threshold:.3f}: стоимость {cost_best}")
print(f"Порог 0.5: стоимость {cost_05}")
print(f"Без модели: стоимость {cost_never}")
```

```python
# Проверка (не изменяйте)
assert total_cost(np.array([1, 0, 1, 0]), np.array([0.9, 0.9, 0.1, 0.1]), 0.5) == COST_FN + COST_FP
assert cost_never == COST_FN * y_test.sum()
assert best_threshold < 0.5, "При дорогих пропусках порог должен быть ниже 0.5"
assert cost_best < cost_05
assert test_pr_auc > 0.8
print("OK")
```
